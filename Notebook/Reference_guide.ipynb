
# Regression Models Reference Guide

The following regression models are used to predict continuous numerical values. Each model has different strengths and weaknesses, so multiple models are typically trained and compared to identify the best-performing solution.

---

## Linear Regression

**Purpose:** Establishes a linear relationship between the input features and the target variable.

**Advantages:**
- Simple and fast
- Easy to interpret
- Good baseline model

**Disadvantages:**
- Assumes linear relationships
- Sensitive to outliers

**Use Case:** Baseline regression model for comparison.

---

## Ridge Regression

**Purpose:** Linear Regression with L2 regularization.

**Advantages:**
- Reduces overfitting
- Handles multicollinearity well
- More stable than Linear Regression

**Disadvantages:**
- Does not perform feature selection

**Use Case:** When Linear Regression is overfitting.

---

## Lasso Regression

**Purpose:** Linear Regression with L1 regularization.

**Advantages:**
- Reduces overfitting
- Automatically performs feature selection
- Can eliminate irrelevant features

**Disadvantages:**
- May remove useful features if regularization is too strong

**Use Case:** Datasets with many features.

---

## K-Nearest Neighbors (KNN) Regressor

**Purpose:** Predicts values based on the average target value of the nearest neighbors.

**Advantages:**
- Simple and intuitive
- No assumptions about data distribution

**Disadvantages:**
- Computationally expensive on large datasets
- Requires feature scaling

**Use Case:** Small datasets with local patterns.

---

## Decision Tree Regressor

**Purpose:** Creates decision rules to make predictions.

**Advantages:**
- Easy to understand
- Handles non-linear relationships
- No feature scaling required

**Disadvantages:**
- Prone to overfitting

**Use Case:** Capturing complex decision boundaries.

---

## Random Forest Regressor

**Purpose:** Combines multiple Decision Trees and averages their predictions.

**Advantages:**
- High accuracy
- Reduces overfitting
- Handles non-linear data well

**Disadvantages:**
- Less interpretable
- Slower than Linear Regression

**Use Case:** Strong general-purpose regression model.

---

## AdaBoost Regressor

**Purpose:** Sequentially builds weak learners that focus on correcting previous errors.

**Advantages:**
- Improves weak models
- Can achieve high prediction accuracy

**Disadvantages:**
- Sensitive to noisy data and outliers

**Use Case:** Improving prediction performance through boosting.

---

## Support Vector Regressor (SVR)

**Purpose:** Uses Support Vector Machines for regression tasks.

**Advantages:**
- Effective for non-linear relationships
- Works well on medium-sized datasets

**Disadvantages:**
- Slow on large datasets
- Requires feature scaling

**Use Case:** Complex regression problems with limited data.

---

## XGBoost Regressor

**Purpose:** Advanced Gradient Boosting algorithm that sequentially improves prediction accuracy.

**Advantages:**
- Highly accurate
- Industry standard for tabular data
- Handles missing values well
- Fast training

**Disadvantages:**
- More hyperparameters to tune
- Less interpretable than simpler models

**Use Case:** Customer Churn, Fraud Detection, Risk Analysis, Kaggle Competitions.

---

## CatBoost Regressor

**Purpose:** Gradient Boosting algorithm optimized for categorical and tabular datasets.

**Advantages:**
- Excellent performance on tabular data
- Handles categorical features efficiently
- Requires less preprocessing
- Strong default settings

**Disadvantages:**
- Can be slower than LightGBM in some cases

**Use Case:** Banking, Insurance, Customer Analytics, E-Commerce.

---

# Model Selection Strategy

1. Start with **Linear Regression** as a baseline.
2. Use **Ridge** or **Lasso** if regularization is needed.
3. Try **Decision Tree** for non-linear relationships.
4. Use **Random Forest** for improved tree-based performance.
5. Test **AdaBoost** for boosting-based improvements.
6. Evaluate **SVR** on smaller datasets.
7. Compare against **XGBoost** and **CatBoost** for maximum predictive performance.

---

# Evaluation Metrics

### Mean Absolute Error (MAE)

Measures the average absolute difference between actual and predicted values.

```python
mean_absolute_error(y_true, y_pred)
```

Lower is better.

---

### Mean Squared Error (MSE)

Measures the average squared difference between actual and predicted values.

```python
mean_squared_error(y_true, y_pred)
```

Lower is better.

---

### Root Mean Squared Error (RMSE)

Square root of MSE.

```python
np.sqrt(mean_squared_error(y_true, y_pred))
```

Lower is better.

---

### R² Score

Measures how well the model explains the variability of the target variable.

```python
r2_score(y_true, y_pred)
```

Range:
- 1.0 = Perfect model
- 0.0 = No predictive power
- Negative = Worse than predicting the mean

Higher is better.

---

# Quick Interview Notes

- **Linear Regression** → Simple baseline model.
- **Ridge Regression** → Reduces overfitting using L2 regularization.
- **Lasso Regression** → Performs feature selection using L1 regularization.
- **KNN Regressor** → Predicts using nearest neighbors.
- **Decision Tree** → Rule-based model capable of learning non-linear patterns.
- **Random Forest** → Ensemble of Decision Trees with reduced overfitting.
- **AdaBoost** → Sequential boosting algorithm that focuses on errors.
- **SVR** → Support Vector Machine for regression tasks.
- **XGBoost** → Industry-standard boosting algorithm for tabular data.
- **CatBoost** → Powerful boosting algorithm optimized for categorical features.

**Recommended Workflow:**

Linear Regression → Random Forest → XGBoost → CatBoost → Select Best Model Based on R² and RMSE
